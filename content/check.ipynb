```python
import geopandas as gpd
from libpysal.graph import Graph
from neighbayes.models import SAR

chi = gpd.read_file("data/Chicago-SDOH/Chi-SDOH.shp")
graph = Graph.build_contiguity(chi, rook=False).transform("r")
model = SAR(formula="YPLL_rate ~ HIS_ct", data=chi, W=graph)
model.fit(draws=200, tune=200, chains=2, random_seed=0, progressbar=False)
model.summary(var_names=["rho"], round_to=2)
```
